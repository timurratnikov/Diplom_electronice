```python
import pandas as pd
import numpy as np

features = pd.read_csv('features.csv')
features
```

```python
corr = features.drop(columns='Phone_new').corr()
corr
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(11, 11))
sns.heatmap(corr)
```

```python
thr = 0.8
for col in corr.columns:
    for i in range(len(corr)):
        if corr.columns[i] == col:
            continue
        if corr[col].iloc[i] > thr:
            print(f'{col} | {corr.columns[i]} | {corr[col].iloc[i]}')
```
