```python
import pandas as pd
import numpy as np
import math

df = pd.read_pickle('electronika_clean.pkl')
df
```

```python
cl = pd.read_csv('clustering_results_1.csv')
cl
```

```python
join = df.join(cl[['Phone_new', 'cluster']].set_index('Phone_new'), on='Phone_new')
join['cluster'].value_counts()
```

```python
actions_values = join['Actions'].value_counts().index
actions_set = set()
for action in actions_values:
    values = action.split(',')
    for value in values:
        actions_set.add(value)

for col in list(actions_set):
    join[col] = join['Actions'].apply(lambda x: int(str(x).find(col) != -1))
join
```

```python
join['QuantAction'] = join['Quant'] * join['СкидкаСуммойПроцентомСЛимитом']
action = join[join['TN'] == 'Крупная бытовая техника'][['OrderDate', 'QuantAction']].groupby(['OrderDate']).sum()
action.head(60)
```

```python
cluster_0_sales = join[(join['TN'] == 'Крупная бытовая техника') & (join['cluster'] == 0)]
total_sales = join[join['cluster'] == 0][['Quant', 'OrderDate']].groupby(['OrderDate']).sum()
cluster_0_sales_grouped = cluster_0_sales[['Quant', 'OrderDate']].groupby(['OrderDate']).sum()
cluster_0_sales_grouped_join = cluster_0_sales_grouped.join(total_sales, rsuffix='_total')
cluster_0_sales_grouped_join
```

```python
cluster_5_sales = join[(join['TN'] == 'Крупная бытовая техника') & (join['cluster'] == 5)]
total_sales = join[join['cluster'] == 5][['Quant', 'OrderDate']].groupby(['OrderDate']).sum()
cluster_5_sales_grouped = cluster_5_sales[['Quant', 'OrderDate']].groupby(['OrderDate']).sum()
cluster_5_sales_grouped_join = cluster_5_sales_grouped.join(total_sales, rsuffix='_total')
cluster_5_sales_grouped_join
```
